# 4. Rispondere alle domande
Trasforma le seguenti richiesti in formule per estrarre infromazioni dal dataframe
- Mostrami le vendite maggiori di 10

- Mostrami i dati del 2018

- Mostrami le vendite maggiori di 13 e l'anno è il 2018

- Mostrami tutto TRANNE i casi in cui le vendite sono maggiori di 13 e l'anno è il 2018

- Mostrami i dati dove le vendite divise per 3 sono maggiori di 3

- Mostrami i dipendenti i cui nomi sono alfabeticamente dopo la J



```python
import pandas as pd

sales = pd.DataFrame(
    data={
        "employee": [
            "Katrina",
            "Guanyu",
            "Jan",
            "Roman",
            "Jacqueline",
            "Paola",
            "Esperanza",
            "Alaina",
            "Egweyn",
        ],
        "sales": [14, 17, 6, 12, 8, 3, 7, 15, 5],
        "year": [2018, 2019, 2020, 2018, 2020, 2019, 2019, 2020, 2020],
    }
)
sales
```

```python
sales[sales['sales'] > 10]
```

```python
sales[sales['year'] == 2018]
```

```python
sales_2018 = sales[sales['year'] == 2018]
sales_2018[sales_2018['sales'] > 13]
```

```python
sales_no2018 = sales[(sales['year'] != 2018) | (sales['sales'] <= 13)]
sales_no2018
```

```python
sales[sales['sales'] / 3 > 3]
```

```python
sales.sort_values(by = 'employee').query('employee > "J"')
```

# 5. Analisi di un dataset di video giochi
1. Quanti videogiochi sono stati pubblicati
2. Quali sono i generi più popolari? Mostralo con un bar plot
3. Plotta l'evoluzione del numero di giochi pubblicati nel tempo
4. Prova a riprodurre questo plot 




```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
df.head()
```

```python
print(f' numero di videogiochi venduti : {df.shape[0]}')
```

```python
top_generi = df['Genre'].value_counts().head(6)
x = top_generi.index      
y = top_generi.values

plt.bar(x , height = y)
```

```python
n_pubblicati = df['Year'].value_counts()
x = n_pubblicati.index      
y = n_pubblicati.values

plt.bar(x , height = y)
```

```python
generi_target = ['Action', 'Misc', 'Role-Playing', 'Shooter', 'Sports']
df_grouped = df[df['Genre'].isin(generi_target)].groupby('Genre')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum()

ax = df_grouped.plot(kind='bar', stacked=True, figsize=(10, 6), width=0.6)

# Colori
plt.title('Vendite per regione per genere')
plt.ylabel('Milioni di unità')
plt.xlabel('Genre')
plt.xticks(rotation=90) # Ruota i nomi dei generi in verticale
plt.legend(title=None)  # Rimuove l'eventuale titolo della legenda

# grafico
plt.show()
```
